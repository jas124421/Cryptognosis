```python
# 1. Download Chromosome 22 VCF (Smallest autosome, ~300MB) from official EBI FTP
!wget -c https://ftp.1000genomes.ebi.ac.uk/vol1/ftp/release/20130502/ALL.chr22.phase3_shapeit2_mvncall_integrated_v5b.20130502.genotypes.vcf.gz
!wget -c https://ftp.1000genomes.ebi.ac.uk/vol1/ftp/release/20130502/integrated_call_samples_v3.20130502.ALL.panel
```

```python
# 2. Convert VCF directly to PLINK binary (.bed/.bim/.fam) using PLINK2
!./plink2 --vcf ALL.chr22.phase3_shapeit2_mvncall_integrated_v5b.20130502.genotypes.vcf.gz \
          --make-bed \
          --out chr22_phase3
```

```python
!./plink2 --vcf ALL.chr22.phase3_shapeit2_mvncall_integrated_v5b.20130502.genotypes.vcf.gz \
          --max-alleles 2 \
          --snps-only \
          --make-bed \
          --out chr22_phase3
```

```python
# Filter MAF > 0.05 and select independent SNPs
!./plink2 --bfile chr22_phase3 \
          --maf 0.05 \
          --indep-pairwise 50 5 0.2 \
          --out pruned_snps

# Extract the pruned SNP set into final clean files
!./plink2 --bfile chr22_phase3 \
          --extract pruned_snps.prune.in \
          --make-bed \
          --out chr22_clean
```
